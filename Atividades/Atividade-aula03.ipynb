# **Exercício 2 — Construindo uma cadeia**
Crie:

```python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
```

Conecte:

```python
A → B → C → None
```

Depois percorra a estrutura e produza:

```python
A
B
C
```

```python
#Resposta

class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None


n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

# Conectando os nós
n1.proximo = n2
n2.proximo = n3
n3.proximo = None

# Percorrendo a estrutura
atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
```

# **Exercício 3 — Depuração**
Analise:

```python
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
```

Perguntas:

- Qual é o problema?
a variável atual nunca é atualizada.
- Por que o programa não termina?
Porque atual continua apontando para n1 imprimindo 10 eternamente.
- Qual linha deve ser acrescentada?
algo como atual = atual.proximo

- Qual será a saída depois da correção?
10
20
30
